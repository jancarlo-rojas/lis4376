### 4. Return Execution Time: %%time
*Note:* %%time is 'cell magic' and must be first line in Jupyter notebook cell

```python
%%time
import random
for i in range(0,1000000):
    random.random

# display time to generate 1000000 random numbers
# note: placing these comments above code generates and error
```
