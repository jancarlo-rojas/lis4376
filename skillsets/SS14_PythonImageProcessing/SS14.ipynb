```python
# imports
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## get image - imread()

```python
img = cv2.imread('watch.jpg')

#check if image is found
if img is None:
    print ("Check file path")
```

## Get image info

```python
print(type(img))
print (img.shape)
print(img.dtype)
```

```python
# image dimensions
dimensions = img.shape

#get height, width, number of channels
height= img.shape[0]
width= img.shape[1]
channels = img.shape[2]

#print
print('Image Dimension: ',dimensions)
print('Image Height: ',height)
print('Image Width: ',width)
print('Number of Channels: ',channels)
```

## Show image - imshow()

```python
#display image in jupyter cell
plt.imshow(img)
plt.title(' my BGR pic ')
plt.show()
```

## Image Conversions

```python
# convert to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title('my RGB pic')
plt.show()
```

## Save Image - imwrite()

```python
cv2.imwrite('watch_new.png', img)
```

## More Image Processing

```python
myImg = cv2.imread('cat-grayscale.jpg')

resized = cv2.resize(myImg, (128, 128), interpolation = cv2.INTER_AREA) # Here, shrinking image
mynewimg = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY) # Returns array of an image
```

```python
# get info
print(mynewimg.shape)
print(type(mynewimg))
print(type(mynewimg.shape))
```

```python
#display numpy array contents
print(mynewimg[:15, :15])
```

```python
#display new img
print(mynewimg.shape)
plt.imshow(mynewimg, cmap='gray')
plt.colorbar()
plt.show()
```

```python
# save as text file
# Numpy savetxt() function: Saves Numpy array to text file
np.savetxt('mycat.csv', mynewimg, fmt='%.1f', delimiter=',')
```
