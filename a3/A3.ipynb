## Pandas pivot() function
*Parameters*:
- index: column to use to make new DataFrames index. if none uses existing index
- columns: column to use to make new DataFrames columns
- values: column(s) to use for populating new DataFrames values. If not specified all remaining columns used
- returns: reshaped DataFrame
- exception: ValueError raised if there are any duplicate indexes

## Shape Data - using melt (long data)
- Each variable is a colimn
- Each observation is a row
- can access variables in dataset by name

### Parameters:
- id_vars: column to use as a identifier variables
- value_vars: column to unpivot, if not specified uses all columns not set as id_vars
- var_name: name to use for variable column, if none uses frame.columns.name or 'variable'
- value_name: name to use for 'value' column
- col_level: if columns multiindex the use this level to melt
- ignore_index: if true, original index ignored. if false original index retained. index labels will be repeated as necessary
- returns: reshaped DataFrame

```python
# 38. melt following dataFrame: 'mortality_wide'
mortality_long = mortality_wide.melt(
    id_vars='Year',
    value_vars=['01-04 Years', '05-09 Years'],
    var_name='AgeGroup',
    value_name='DeathRate')
mortality_long.head(4)
```

```python
# 39. using pandas option_context() function, display 6 rows (first and last 3) and all cols
with pd.option_context(
    'display.max_rows', 6,
        'display.max_columns', None):
    print(mortality_long)
```
